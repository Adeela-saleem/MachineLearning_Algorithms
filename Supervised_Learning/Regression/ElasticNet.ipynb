Multicollinearity: Jab aapke dataset mein do ya zyada features ek doosre se bohot zyada milte-julte (correlated) hon, toh Lasso randomly kisi ek feature ko pick karta hai aur baki ko zero kar deta hai. ElasticNet dono features ke weights ko balance karke retain karta hai.


```python
from sklearn.datasets import load_diabetes
X,y = load_diabetes(return_X_y=True)

```

```python
from sklearn.model_selection import train_test_split

X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.2,random_state=2)
```

```python
from sklearn.linear_model import LinearRegression,Ridge,Lasso,ElasticNet

```

```python
from sklearn.metrics import r2_score
```

```python
# Linear Regression
reg = LinearRegression()
reg.fit(X_train,y_train)
y_pred = reg.predict(X_test)
r2_score(y_test,y_pred)
```

```python
# Ridge
reg = Ridge(alpha=0.1)
reg.fit(X_train,y_train)
y_pred = reg.predict(X_test)
r2_score(y_test,y_pred)
```

```python
# Lasso
reg = Lasso(alpha=0.01)
reg.fit(X_train,y_train)
y_pred = reg.predict(X_test)
r2_score(y_test,y_pred)
```

```python
# ElasticNet
reg = ElasticNet(alpha=0.005,l1_ratio=0.9)
reg.fit(X_train,y_train)
y_pred = reg.predict(X_test)
r2_score(y_test,y_pred)
```
