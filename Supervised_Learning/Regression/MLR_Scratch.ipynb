### **Why Load Datasets Directly as NumPy Arrays (`return_X_y=True`) Instead of a Pandas DataFrame?**

1. **Efficiency for Mathematical Operations:**
* When implementing ML algorithms from scratch (e.g., Multiple Linear Regression), operations like matrix multiplication ($X^T X$), transpositions, and inversions are executed much faster and more directly using **NumPy arrays**.


2. **Scikit-Learn Compatibility:**
* Scikit-Learn models (like `LinearRegression().fit(X, y)`) natively process NumPy arrays. Passing a DataFrame internally converts it to a NumPy array anyway.


3. **No Data Cleaning Needed:**
* Pre-packaged benchmark datasets (like `load_diabetes()`) are already clean, missing-value-free, scaled, and purely numerical. Therefore, Pandas data-wrangling features (`df.fillna()`, `df.describe()`) are not required.



> **When to use Pandas instead?** Use a Pandas DataFrame when performing Exploratory Data Analysis (EDA), inspecting column names, handling categorical data, or cleaning raw real-world datasets.

* **Model Limitation:** Multiple Linear Regression assumes a strictly linear relationship between input features and target variables. Real-world medical variables often exhibit non-linear interactions that linear models cannot fully capture.
* **Benchmarking:** A score of **$\sim 44\%$** is standard for a simple Linear Regression model on this dataset. Advanced techniques like regularization (Ridge/Lasso) or non-linear algorithms (e.g., Decision Trees, Random Forest) are required to boost accuracy.

### **2. Interpretation of Regression Coefficients (`reg.coef_`)**

Linear Regression predicts values using the formula:

$$y = \text{Intercept} + \beta_1 X_1 + \beta_2 X_2 + \dots + \beta_{10} X_{10}$$

The `reg.coef_` array contains 10 values ($\beta_1 \dots \beta_{10}$), corresponding to the **10 features** in the dataset:

* **Positive Coefficients (e.g., $+861.12$, $+516.69$):** Represent a **direct relationship**. As these feature values increase, disease progression increases.
* **Negative Coefficients (e.g., $-895.55$, $-205.45$):** Represent an **inverse relationship**. As these feature values increase, disease progression decreases.
* **Magnitude (Absolute Value):** Indicates feature importance. Features with larger numerical values (e.g., $-895.55$ and $+861.12$) exert the strongest influence on the model's predictions, whereas smaller values (e.g., $-9.15$ or $+52.42$) have minimal impact.

# Making our own Linear Regression Class

### **Custom Multiple Linear Regression (`ownClass`) Implementation Summary**

1. **Constructor (`__init__`):**
* Initializes `self.coef_` (weights/slopes) and `self.intercept_` (bias/$Y$-intercept) as `None`.
* Follows Scikit-Learn convention by using trailing underscores for parameters learned during model training.


2. **Inserting the Bias/Dummy Column (`np.insert`):**
* **Formula:** Multiple Linear Regression equation is $y = \beta_0(1) + \beta_1 X_1 + \dots + \beta_n X_n$.
* **Purpose:** Adds a column of ones ($1\text{s}$) at index `0` of `X_train` (`axis=1`). This allows the intercept term ($\beta_0$) to be computed seamlessly alongside feature coefficients ($\beta_1 \dots \beta_n$) using matrix operations (Normal Equation: $\boldsymbol{\beta} = (\mathbf{X}^T \mathbf{X})^{-1} \mathbf{X}^T \mathbf{y}$).

```python
class ownClass:

  def __init__(self):
      self.coef_=None
      self.intercept_=None
  def fit(self,X_train,y_train):
    X_train=np.insert(X_train,0,1,axis=1)

    #calculate coeffs
    betas=np.linalg.inv(np.dot(X_train.T,X_train)).dot(X_train.T).dot(y_train)
    self.intercept_=betas[0]
    self.coef_=betas[1:]

  def predict(self,X_test):
    y_pred=np.dot(X_test,self.coef_)+self.intercept_
    return y_pred


```

### **Mathematical Formulation of OLS (Normal Equation)**

1. **Normal Equation Computation:**
   * Uses the closed-form formula $\boldsymbol{\beta} = (\mathbf{X}^T \mathbf{X})^{-1} \mathbf{X}^T \mathbf{y}$ via `np.linalg.inv()` and `.dot()` to find optimal parameters in a single step without Gradient Descent.

2. **Extracting Intercept and Coefficients:**
   * **`self.intercept_ = betas[0]`**: The first element represents the bias/intercept ($\beta_0$) corresponding to the inserted dummy column of ones.
   * **`self.coef_ = betas[1:]`**: The remaining values represent feature weights ($\beta_1 \dots \beta_n$).

```python
lr=ownClass()
```

```python
lr.fit(X_train,y_train)
```
