```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python

from sklearn.datasets import load_diabetes

data=load_diabetes()
```

```python
print(data.DESCR)
```

```python
X=data.data
```

```python
y=data.target
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.2,random_state=45)
```

```python

from sklearn.linear_model import LinearRegression
L=LinearRegression()

```

```python
L.fit(X_train,y_train)
```

```python
print(L.coef_)
print(L.intercept_)
```

```python
y_pred=L.predict(X_test)
```

```python
from sklearn.metrics import r2_score,mean_squared_error

print("R2 score",r2_score(y_test,y_pred))
print("RMSE",np.sqrt(mean_squared_error(y_test,y_pred)))
```

```python

from sklearn.linear_model import Ridge
R=Ridge(alpha=100000)
```

```python
R.fit(X_train,y_train)
```

```python
print(R.coef_)
print(R.intercept_)
```

```python
y_pred1=R.predict(X_test)
```

```python
print("R2 score",r2_score(y_test,y_pred1))
print("RMSE",np.sqrt(mean_squared_error(y_test,y_pred1)))
```

### **Ridge Regularization Behavior Analysis**

1. **Impact of Over-Regularization ($\alpha = 100,000$):**
   * Setting an excessively high penalty term ($\alpha$) forces all regression coefficients ($\boldsymbol{\beta}$) close to zero ($\boldsymbol{\beta} \approx \mathbf{0}$).
   * **Learned Coefficients:** Drops from large values (e.g., $-970.79$) to near-zero ($0.0023$).

2. **Performance Collapse (Underfitting):**
   * Baseline Linear Regression $R^2$: **~0.518**
   * Ridge Regression ($\alpha=100,000$) $R^2$: **~ -0.0004**
   * When $\alpha \to \infty$, the model simplifies to predicting only the target mean ($\hat{y} \approx \bar{y}$), leading to standard severe underfitting.

3. **Key Rule:**
   * $\alpha$ needs proper tuning (e.g., using `RidgeCV` or GridSearch over $10^{-3} \dots 10^2$) to balance bias-variance tradeoff without crushing model capacity.

```python
from sklearn.linear_model import Ridge
from sklearn.metrics import r2_score

alphas = [0.0001, 0.001, 0.01, 0.1, 1, 10, 100, 1000]

for a in alphas:
    r = Ridge(alpha=a)
    r.fit(X_train, y_train)
    y_pred = r.predict(X_test)
    print(f"Alpha: {a:7} | R2 Score: {r2_score(y_test, y_pred):.5f}")
```

### **Optimal Alpha Selection Findings**

1. **Best Hyperparameter Value:**
   * $\alpha = 0.01$ yields the optimal test $R^2$ score of **~0.5231**, outperforming ordinary Linear Regression (**~0.5188**).

2. **Regularization Tradeoff:**
   * **Small Penalty ($\alpha \le 0.1$):** Properly controls variance without inflating bias.
   * **Large Penalty ($\alpha \ge 1$):** Forces severe underfitting, causing $R^2$ score to drop drastically towards 0.

 Practical Machine Learning WorkflowReal-world ML pipeline mein standard workflow yeh hota hai:Step 1: Pehle simple Linear Regression as a baseline chalayein.Step 2: Training score aur Testing score ko compare karein:Agar Train Score $\approx$ Test Score $\rightarrow$ Ridge ki zaroorat nahi hai.Agar Train Score $\gg$ Test Score (Overfitting) $\rightarrow$ Ridge/Lasso apply karein.Step 3: Automated cross-validation tool (jaise RidgeCV ya GridSearchCV) se optimal penalty value $\alpha$ dhoond kar model finalize karein.
